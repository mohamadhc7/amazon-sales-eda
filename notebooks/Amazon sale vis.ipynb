```python
data["ship-state"] = data["ship-state"].str.upper()
```

```python
data["ship-state"] = data["ship-state"].replace("RAJSTHAN", "RAJASTHAN")
```

```python
data["ship-state"].unique()
```

```python
data['ship-state'].value_counts()
```

```python
data["ship-state"] = data["ship-state"].replace("NEW DELHI", "DELHI")
```

```python
data.isnull().sum()
```

```python
data=data.dropna(subset=["ship-state"])
```

Ship-state cleaning:

Casing issue: State names were inconsistently formatted (mix of uppercase and lowercase), so I converted the entire column to uppercase for    consistency.

Misspelling: Found "RAJSTHAN" misspelled instead of "RAJASTHAN," so I replaced it with the correct spelling.

Duplicate naming: "NEW DELHI" was being counted separately from "DELHI," so I merged them into "DELHI" for a more accurate count.

Missing values: 33 rows had no ship-state value, so I dropped those rows since they were a small fraction of the dataset.

Ship-city cleaning:

Casing issue: City names were inconsistently formatted (mix of uppercase and lowercase), so I converted the entire column to uppercase for consistency.

Duplicate naming: "BANGALORE" was being counted separately from "BENGALURU" (the city's official renamed version), so I merged them into "BENGALURU" for a more accurate count.

Chart 1: Expedited shipping makes up the majority of orders (~69%), while Standard accounts for the remaining ~31%.

Chart 2: Set and kurta are the two most shipped categories, together making up about 78% of all orders, while no other category exceeds ~12%.

Chart 3: Maharashtra has the highest number of shipped orders (~17%) among all 31 states/UTs, though no single state holds a majority.

Chart 4: Bengaluru receives the highest number of orders nationally (~10%), ahead of other major cities, but does not account for a majority.


Given that most deliveries are Expedited, a business might consider improving Standard delivery speed to reduce the gap between the two options.

Given the dominance of Set and kurta, a business might consider running promotions on lower-performing categories (Saree, Dupatta, Bottom) to boost their sales.

Given Maharashtra's lead in order volume, a business might consider prioritizing warehouse or fulfillment capacity there to maintain fast delivery as demand grows.

Given Bengaluru's high order volume, a business might consider setting up a local distribution hub to reduce delivery times in that city.
